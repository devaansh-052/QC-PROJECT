Run `python scripts/run_eda.py` from the project root first. This notebook reads only saved aggregates; it never reloads the full CSV. Select an explicit report path when citing a run. Balance figures are diagnostic only.


```python
from pathlib import Path
import json
import pandas as pd
from quantum_fraud.data.eda import summarize_eda  # authoritative aggregation implementation
root = Path.cwd() if (Path.cwd() / "configs").exists() else Path.cwd().parent
reports = sorted((root / "results/metrics").glob("eda_*/report.json"))
if not reports:
    raise FileNotFoundError("Run scripts/run_eda.py first")
report_path = reports[-1]  # replace with an explicit report path for a fixed run
report = json.loads(report_path.read_text())
print(report_path)
print(json.dumps(report["validation"], indent=2))
```

```python
for name in report["tables"]:
    print("\n" + name)
    print(pd.read_csv(report_path.parent / name).head(12).to_string(index=False))
```

```python
import matplotlib.pyplot as plt
for name in report["figures"]:
    if name.endswith(".png"):
        fig, ax = plt.subplots(figsize=(12, 5))
        ax.imshow(plt.imread(Path(report["figures_dir"]) / name))
        ax.axis("off")
        plt.show()
        plt.close(fig)
```
