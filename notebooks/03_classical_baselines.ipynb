# Classical baseline results
Run `scripts/train_classical.py --prepared PATH` first. This notebook reads completed result metadata only. Scores are preliminary, single-seed and uncalibrated.


```python
from pathlib import Path
import json
import pandas as pd
root = Path.cwd() if (Path.cwd()/"configs").exists() else Path.cwd().parent
runs = sorted((root/"results/metrics").glob("classical_*/results.json"))
if not runs:
    raise FileNotFoundError("Complete a classical run first")
run_path = runs[-1]  # select an explicit run when citing results
result = json.loads(run_path.read_text())
print(run_path)
table = pd.DataFrame([{ "model": r["model"], **{k:v for k,v in r["test"].items() if k != "confusion_matrix"}, "training_seconds":r["training_seconds"]} for r in result["models"]])
print(table.to_string(index=False))
```

See `docs/CLASSICAL_BASELINES.md` for threshold selection, timings and limitations. Future QSVC comparisons require a matched kernel SVM, not simply this linear SVM.
