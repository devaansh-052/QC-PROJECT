# Feature-reduction experiments
Read `docs/FEATURE_REDUCTION.md` first. This notebook only loads saved aggregate results, never training data. The 8-feature case is unsupported by the frozen representation.


```python
from pathlib import Path
import json
import pandas as pd
root = Path.cwd() if (Path.cwd()/"configs").exists() else Path.cwd().parent
reports = sorted((root/"results/metrics").glob("features_*/results.json"))
if not reports: raise FileNotFoundError("Run scripts/reduce_features.py first")
report_path = reports[-1]  # choose an explicit report path when citing results
report = json.loads(report_path.read_text())
print(report_path)
print("Unsupported dimensions:", report["unsupported_dimensions"])
print(pd.DataFrame([{ "method": c["method"], "dimension":c["dimension"], "features":c["features"], "test_average_precision":c["test"]["pr_auc_average_precision"], "test_f1":c["test"]["f1"]} for c in report["controls"]]).to_string(index=False))
print("PCA variance ratios:", report["pca_explained_variance_ratio"])
```
